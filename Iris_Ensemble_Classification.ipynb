# Visualizing Features
**Class features visualization** using **scatter plots** is a foundational technique in exploratory data analysis (EDA) and machine learning, used to understand the relationship between numerical features, identify data clusters, and detect outliers. In classification problems, scatter plots are particularly useful for visualizing the separation between different classes by mapping class labels to colors or shapes.

````python
# Create a scatter plot matrix of the features using seaborn
import seaborn as sns
import matplotlib.pyplot as plt

# Create a scatter plot of the two features 'PetalLengthCm' and 'PetalWidthCm'
# colored by the target variable 'Species'
sns.scatterplot(data=dataset, x='PetalLengthCm', y='PetalWidthCm', hue='Species', 
                palette='Set1', style='Species', markers=['o', 's', '^'], s=100)

# Show the plot
plt.title('Scatter Plot of Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.legend(title='Species')
plt.show()
````

**Box and whisker plots** (box plots) are powerful visualization tools for analyzing class features (attributes) by displaying the distribution, central tendency, spread, and skewness of numerical data. They are particularly effective for comparing characteristics across multiple categories, such as comparing test scores between classes or analyzing features within a dataset.

**Key Components Visualized**
* **Median (Q2)**: The middle value of the data, indicated by a line inside the box.
* **Interquartile Range (IQR)**: The box, representing the middle 50% of data (Q1 to Q3).
* **Whiskers**: Lines extending to the minimum and maximum non-outlier values, showing the range of data.
* **Outliers**: Individual dots beyond the whiskers, highlighting unusual observations.

````python
# Create a box and whisker plot to visualize the distribution of 'PetalLengthCm' for each species
sns.boxplot(x='Species', y='PetalLengthCm', data=dataset, palette='Set2', hue='Species')
plt.title('Box Plot of Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.show()
````

**Density plots** also known as *Kernel Density Estimation plots* are a powerful visualization tool used to show the probability density function of a continuous, numerical variable. They act as a smoothed version of a histogram, making them ideal for identifying patterns such as skewness, modality (peaks), and clusters across different classes in a dataset.

````python
# Create a Density Plot to visualize the distribution of 'PetalLengthCm' for each species
sns.kdeplot(data=dataset, x='PetalLengthCm', hue='Species', fill=True, palette='Set1')
plt.title('Density Plot of Petal Length by Species')
plt.xlabel('Petal Length')
plt.ylabel('Density')
plt.show()
````

**Violin plots** are powerful data visualization tools used to compare the distribution, density, and variability of numerical features across different classes or categories. They are superior to box plots when data is multimodal (has multiple peaks) or skewed, as they display the full probability density of the data.

````python
# Create a Violin Plot to visualize the distribution of 'PetalLengthCm' for each species
sns.violinplot(x='Species', y='PetalLengthCm', data=dataset, palette='Set1', hue='Species')
plt.title('Violin Plot of Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.show()
````

# Assigning Features and Target variables from the dataset
**Features** also known as *predictors* are individual independent variables that act as input data in the classifier model.

**Targets** are the final output data trying to predict.

````python
# Assigned data variables as features
features = dataset.drop("Species", axis=1)

# Assigned label variables as targets
targets = dataset["Species"]

# Display 10 random features dataframe
features.sample(10).style.hide(axis="index")
````

````python
# Display 10 random targets dataframe
targets.sample(10)
````

# Splitting the Dataset into Training, Testing, and Validation Set
Splitting a dataset is a crucial step in machine learning to evaluate model performance and prevent overfitting. Typically, the data is divided into three sets:
- **Training Set**: Used to train the model.
- **Validation Set**: Used to tune hyperparameters and select the best model.
- **Testing Set**: Used to assess the final model performance.

A common approach is to first split the data into training and testing sets, then further split the training set into training and validation sets. This ensures that the validation and test sets remain unseen during training and hyperparameter tuning.

<img src="https://cdn.imweb.me/upload/S202105076e626e4618b27/0f0c2df8f341c.png" style="width:70%; height:auto;">

````python
# Include Train test set split function from scikit-learn
from sklearn.model_selection import train_test_split

# Split data set to training set (80%) and test set (20%)
X_train, X_test, y_train, y_test = train_test_split(features.astype('float32'), targets, test_size=0.20, shuffle=True, random_state=43, stratify=targets)

# Create a dataframe that holds the dataset shape (number of samples and features) for training and testing sets
dataset_shape = pd.DataFrame({
    'Dataset': ['Dataset', 'Training Set', 'Testing Set'],
    'Samples': [features.shape[0], X_train.shape[0], X_test.shape[0]],
    'Features': [features.shape[1], X_train.shape[1], X_test.shape[1]],
    'Target Classes': [len(set(targets)), len(set(y_train)), len(set(y_test))]
})

# Display the dataset shape
dataset_shape.style.hide(axis='index')
````

# Train the Classifier Model
* **Decision Tree**: Is one of the simplest classification algorithms you can use in ML. It categorizes data by recursively splitting it based on feature-driven decision rules. It mimics human decision-making by using a flowchart-like structure to reach a final classification.
* **RandomForestClassifier**: Builds multiple decision trees on random subsets of data and averages them to improve accuracy and control overfitting.
* **ExtraTreesClassifier**: Similar to Random Forest but uses "extremely randomized trees" picking random thresholds for features rather than searching for the best threshold, which speeds up training and increases randomization.

````python
# Import Decision Tree Algorithm from Scikit-Learn
from sklearn.tree import DecisionTreeClassifier

# Create Decision Tree classifer object
dt_clf = DecisionTreeClassifier(random_state=42, max_depth=10)

# Train the Decision Tree Classifer Model
dt_clf.fit(X_train.astype('float32'), y_train)
````

````python
# Import Random Forest Classifier from scikit-learn
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest Classifier object
rf_clf = RandomForestClassifier(n_estimators=30, random_state=42, max_depth=10)

# Train the Random Forest Classifier Model
rf_clf.fit(X_train.astype('float32'), y_train)
````

````python
# Import Extra Trees Classifier from scikit-learn
from sklearn.ensemble import ExtraTreesClassifier

# Create Extra Trees Classifier object
et_clf = ExtraTreesClassifier(n_estimators=30, random_state=42, max_depth=10)

# Train the Extra Trees Classifier Model
et_clf.fit(X_train.astype('float32'), y_train)
````

# Evaluating the Performance of the Classifier Model
**Classification metrics** evaluate the performance of machine learning models that categorize data into distinct classes. Key metrics include accuracy (overall correctness), precision (correct positive predictions), recall/sensitivity (finding all positive samples), and the F1-score (harmonic mean of precision and recall). The choice of metric depends on class balance.

### Key Classification Metrics:
* **Accuracy**: Percentage of correct predictions out of all predictions. It is best for balanced datasets.
* **Precision**: Proportion of true positive predictions among all positive predictions. Use this when the cost of false positives is high.
* **Recall** (Sensitivity): Proportion of true positive predictions among all actual positive instances. 
* **F1-Score**: The harmonic mean of precision and recall, balancing both measures. Ideal for imbalanced data.


````python
# Print the performance metrics such as accuracy, precision, recall, and F1-score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import seaborn as sns
import matplotlib.pyplot as plt

# Create a dataframe to hold the classification metrics for each model
metrics_df = pd.DataFrame({
    'Model': ['Decision Tree', 'Random Forest', 'Extra Trees'],
    'Accuracy': [f'{score:.2f}%' for score in [
        accuracy_score(y_test, dt_clf.predict(X_test.astype('float32'))) * 100.00,
        accuracy_score(y_test, rf_clf.predict(X_test.astype('float32'))) * 100.00,
        accuracy_score(y_test, et_clf.predict(X_test.astype('float32'))) * 100.00
    ]],
    'Precision': [f'{score:.2f}%' for score in [
        precision_score(y_test, dt_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,
        precision_score(y_test, rf_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,
        precision_score(y_test, et_clf.predict(X_test.astype('float32')), average='weighted') * 100.00
    ]],
    'Recall': [f'{score:.2f}%' for score in [
        recall_score(y_test, dt_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,
        recall_score(y_test, rf_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,  
        recall_score(y_test, et_clf.predict(X_test.astype('float32')), average='weighted') * 100.00
    ]],
    'F1 Score': [f'{score:.2f}%' for score in [
        f1_score(y_test, dt_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,
        f1_score(y_test, rf_clf.predict(X_test.astype('float32')), average='weighted') * 100.00,
        f1_score(y_test, et_clf.predict(X_test.astype('float32')), average='weighted') * 100.00
    ]]
})

# Display the metrics dataframe
metrics_df.style.hide(axis='index')
````

# Confusion Matrix

A confusion matrix is a useful tool for evaluating the performance of a classification model. It provides a summary of prediction results on a classification problem, showing the number of correct and incorrect predictions broken down by each class.

**Key Points:**
- Each row of the matrix represents the instances in an actual class, while each column represents the instances in a predicted class.
- The diagonal elements represent the number of points for which the predicted label is equal to the true label (correct predictions).
- Off-diagonal elements are those that are mislabeled by the classifier.

**Why use a confusion matrix?**
- To visualize the performance of a classification algorithm.
- To identify types of errors made by the model (e.g., which classes are most often confused).
- To compute important metrics such as accuracy, precision, recall, and F1-score.

````python
# Confusion Matrix and Classification Report
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Compute the confusion matrix for each model
dt_conf_matrix = confusion_matrix(y_test, dt_clf.predict(X_test.astype('float32')))
rf_conf_matrix = confusion_matrix(y_test, rf_clf.predict(X_test.astype('float32')))
et_conf_matrix = confusion_matrix(y_test, et_clf.predict(X_test.astype('float32')))

# Create a list of models and their corresponding predictions
models = ['Decision Tree', 'Random Forest', 'Extra Trees']
predictions = [dt_clf.predict(X_test.astype('float32')), rf_clf.predict(X_test.astype('float32')), et_clf.predict(X_test.astype('float32'))]

# Create subplots for confusion matrices
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot confusion matrices for each model
for i, (model, pred) in enumerate(zip(models, predictions)):
    conf_matrix = confusion_matrix(y_test, pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=conf_matrix, display_labels=dt_clf.classes_)
    disp.plot(ax=axes[i], cmap='Blues', colorbar=False)
    axes[i].set_title(f'{model} Confusion Matrix')
    axes[i].set_xlabel('Predicted Label')
    axes[i].set_ylabel('True Label')
plt.show()
````

## Random Sample Prediction
Below, we select a random sample from the test set and use the trained model to predict its class. The predicted class and the true class are compared to assess the model's performance on individual samples.

````python
# Generate a random number ranges from the test set for inference
import numpy as np
random_index = np.random.randint(0, X_test.shape[0])

# Extract the corresponding test sample and its true label
X_test_sample = X_test[random_index:random_index+1]
y_test_sample = y_test[random_index:random_index+1]

# Perform prediction using sample test sample for each model
dt_pred = dt_clf.predict(X_test_sample)
rf_pred = rf_clf.predict(X_test_sample)
et_pred = et_clf.predict(X_test_sample)

# Compute for probabilities for the sample test sample for each model
dt_prob = dt_clf.predict_proba(X_test_sample)
rf_prob = rf_clf.predict_proba(X_test_sample)
et_prob = et_clf.predict_proba(X_test_sample)

# Get the index of the predicted class
pred_class_idx = np.argmax(dt_prob[0])

# Create a dataframe to hold the test sample and its true label
test_sample_df = pd.DataFrame({'Features': [X_test_sample.astype('float32').to_csv(index=False, header=False).strip()], 
                               'True Label': [y_test_sample.values[0]],
                               'Predicted Label (DT)': [dt_pred[0]],
                               'Predicted Label (RF)': [rf_pred[0]],
                               'Predicted Label (ET)': [et_pred[0]],
                               'Probability (DT)': [float(dt_prob[0, pred_class_idx] * 100)],
                               'Probability (RF)': [float(rf_prob[0, pred_class_idx] * 100)],
                               'Probability (ET)': [float(et_prob[0, pred_class_idx] * 100)]})

# Display the test sample dataframe
test_sample_df.style.hide(axis='index').format({'Probability (DT)': '{:.2f}%', 'Probability (RF)': '{:.2f}%', 'Probability (ET)': '{:.2f}%'})
````

# Visualizing the Model Decision Boundary
Visualizing ensemble multiclass classification decision boundaries involves plotting the high-dimensional model's predictions over a 2D meshgrid using tools like scikit-learn's `DecisionBoundaryDisplay`. This technique divides the feature space into regions representing each class, defining where the model changes predictions to show smoother, non-linear boundaries. For multiclass, this is often a complex, multi-segmented surface. When data has more than tow (2) features, technique like `Principal Component Analysis` (PCA) or `feature selection` is needed to plot boundaries on a 2D plot.

````python
# Import kbest from scikit-learn for feature selection
from sklearn.feature_selection import SelectKBest, f_classif

# Select the top 2 features using ANOVA F-value
selector = SelectKBest(score_func=f_classif, k=2)

# Fit and transform the training and test features using SelectKBest
X_train_selected = selector.fit_transform(X_train, y_train)
X_test_selected = selector.transform(X_test)

# Retrain the models using the selected features
dt_clf_selected = DecisionTreeClassifier(random_state=42, max_depth=10)
dt_clf_selected.fit(X_train_selected, y_train)
rf_clf_selected = RandomForestClassifier(n_estimators=30, random_state=42, max_depth=10)
rf_clf_selected.fit(X_train_selected, y_train)
et_clf_selected = ExtraTreesClassifier(n_estimators=30, random_state=42, max_depth=10)
et_clf_selected.fit(X_train_selected, y_train)

# Import the DecisionBoundaryDisplay from scikit-learn
from sklearn.inspection import DecisionBoundaryDisplay

# Compute the decision boundary for each model using DecisionBoundaryDisplay
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Transform the inference sample to the same selected feature space
X_test_sample_selected = selector.transform(X_test_sample)

for i, (model, clf) in enumerate(zip(models, [dt_clf_selected, rf_clf_selected, et_clf_selected])):
    disp = DecisionBoundaryDisplay.from_estimator(clf, 
                                                  X_test_selected, 
                                                  response_method='predict_proba', 
                                                  plot_method='contourf',
                                                  grid_resolution=100,
                                                  ax=axes[i],                                                   
                                                  alpha=0.7)
    
    # Overlay the test samples on the decision boundary plot
    sns.scatterplot(x=X_test_selected[:, 0], y=X_test_selected[:, 1], hue=y_test, palette='Set1', style=y_test, markers=['o', 's', '^'], s=100, ax=axes[i])
    # Highlight the specific test sample used for inference
    sns.scatterplot(x=X_test_sample_selected[:, 0], y=X_test_sample_selected[:, 1], color='m', marker='X', s=200, label='Test Sample', ax=axes[i])
    axes[i].set_title(f'{model} Decision Boundary')
    axes[i].set_xlabel('Petal Length')
    axes[i].set_ylabel('Petal Width')
    axes[i].legend()
    axes[i].set_xlim(1.0, 7.5)
    axes[i].set_ylim(0.0, 3.0)
plt.show()
````

# Port the Classifier Model to C Code
## emlearn - Machine Learning for Tiny Embedded Systems
```emlearn``` is an open-source Python library that allows converting scikit-learn and Keras models to efficient C code. This makes it easy to deploy to any microcontroller with, while keeping Python-based workflow that is familiar to Machine Learning Engineers.

To install emlearn to convert models in Python, use ```pip install emlearn```

````python
# Convert model using emlearn
import emlearn

# Convert the trained Decision Tree model to C++ code using emlearn
dt_clf_c_model = emlearn.convert(dt_clf, method='inline', dtype='float', return_type='classifier')

# Generate and save the C code to a header file
dt_clf_c_code = dt_clf_c_model.save(file='DecisionTreeClassifier.h', name='DecisionTreeClassifier')

# Print the generated C code
print(dt_clf_c_code)
````

````python
# Convert the trained Random Forest model to C++ code using emlearn
rf_clf_c_model = emlearn.convert(rf_clf, method='inline', dtype='float', return_type='classifier')

# Generate and save the C code to a header file
rf_clf_c_code = rf_clf_c_model.save(file='RandomForestClassifier.h', name='RandomForestClassifier')

# Print the generated C code
print(rf_clf_c_code)
````

````python
# Convert the trained Extra Trees model to C++ code using emlearn
et_clf_c_model = emlearn.convert(et_clf, method='inline', dtype='double', return_type='classifier')

# Generate and save the C code to a header file
et_clf_c_code = et_clf_c_model.save(file='ExtraTreesClassifier.h', name='ExtraTreesClassifier')

# Print the generated C code
print(et_clf_c_code)
````
